```python
# Positional arguements

def user_info(name, age, city):
    '''This function prints name, age, and city
    from an arguement provided to the function.'''

    print("{} is {} years old, from {}".format(name, age, city))

user_info("Jane", 65, "Oklahoma City")
user_info(35, "new")
```

```python
# keyword Arguements 

def user_info(name, age=0, city='Tucson'):
    '''This function prints name, age, and city
    from an arguement provided to the function.'''

    print("{} is {} years old, from {}".format(name, age, city))

user_info("Jane", 65, "Oklahoma City")
user_info("Mike")
user_info(name='Kadeem', age= 45)
```

```python
# *args & **kwargs
'''*args:
contents: allows for unlimited variables to be
passed into a function without defining them ahead of time
'''
def add(*args):
    print(sum(args))

add(2,3,5)
add(2,4,6,8,10)

'''kwargs
contents: allows for unlimited keyword arguements to be passed
into a function without defining them ahead of time'''

def application(**kwargs):
    print(**kwargs)

application(name='Jess', email="mail@mail.com")
application(name="Susan", surname="Johnson", age=42)


# Combining arg types
'''All three arguement types can be used in a single function.
They must be used in order: formal positional arguements, *args, **kwargs'''

def application(fname, lname, email, company, *args, **kwargs):
    print("{} {} works at {}. Her email is {}.".format(fname, lname, company, email))

application("Jess", "Ingrass", "mail@mail.com", "SimpleCode.org", 75000, hire_date= "Sept 2011")
```
