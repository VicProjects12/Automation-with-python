```python
# mini Project

contacts = {
    "Alex": {'phone': '222-222-2222', 'email': 'alex@hotmail.com'},
    "Bob": {'phone': '223-223-2223', 'email': 'bob@gmail.com'},
    "Charles": {'phone': '224-224-2224', 'email': 'cja@gmail.com'}
   
 }
print(contacts)
```

```python
contacts
print(contacts.values())
```

```python
print(contacts.items())
```

```python
for key in contacts.items():
    print(key)
```

```python
for name, info in contacts.items():
    print(name)
    print(info)
```

```python
for name, info in contacts.items():
    print('Name:', name)
    print("phone:", info["phone"])
    print("Email:", info["email"])
    print()
```
